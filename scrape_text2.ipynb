```python
from bs4 import BeautifulSoup
import requests
import sys
```

```python
# desktop user-agent
USER_AGENT = "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:65.0) Gecko/20100101 Firefox/65.0"

headers = {"user-agent": USER_AGENT}
r = requests.get("https://www.geeksforgeeks.org/artificial-intelligence/difference-between-machine-learning-and-deep-learning/", headers=headers)

print("Status code: ", r.status_code)
```

```python
if r.status_code == 200:
    # resp = requests.get(URL, headers=headers)
    soup = BeautifulSoup(r.content, "html.parser")
else:
    print("Scrapping is not possible")
    sys.exit(0)
```

```python
# Output file
out_filename = "test_t.txt"

# Extract title
title_tag = soup.find("h1")
title = title_tag.get_text(strip=True) if title_tag else "No Title"

# Use a  reliable selector
content_div = soup.find("div", {"class": "text"})

if not content_div:
    print("Content not found. Website structure may have changed.")
    sys.exit(0)

# Extract all text inside article
paragraphs = content_div.find_all(["p", "li"])

article_text = ""

for p in paragraphs:
    text = p.get_text(strip=True)
    if text:
        article_text += text + "\n"

# Write to file
with open(out_filename, "w", encoding="utf-8") as f:
    f.write(title + "\n\n")
    f.write(article_text)

print(f"Content saved to {out_filename}")

f.close() # Close the file

```
